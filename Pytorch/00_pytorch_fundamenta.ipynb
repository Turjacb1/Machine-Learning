# Manipulating tensors (tensor operations)

```python
#ADDITION
tensor = torch.tensor([1,2,3])
print(torch.add(tensor, 10))
#tensor +10
```

```python
## Element-wise matrix multiplication
tensor1 = torch.tensor([1, 2, 3])
tensor2 = torch.tensor([4, 5, 6])

print(tensor1 * tensor2)
```

```python
#Matrix Multiplication
tensor1 = torch.tensor([[1, 2],
                        [3, 4]])

tensor2 = torch.tensor([[5, 6],
                        [7, 8]])


print(tensor1 @ tensor2)
```

```python
#torch.matmul()

tensor1 = torch.tensor([[1, 2],
                        [3, 4]])

tensor2 = torch.tensor([[5, 6],
                        [7, 8]])

print(torch.matmul(tensor1, tensor2))
```

```python
# Matrix multiplication by hand
# (avoid doing operations with for loops at all cost, they are computationally expensive)
import torch

tensor = torch.tensor([1, 2, 3])

value = 0

for i in range(len(tensor)):
    value += tensor[i] * tensor[i]

print(value)
```

# Reshaping, stacking, squeezing and unsqueezing




#Reshaping - changes the tensor shape

```python
import torch

x = torch.arange(1, 7)

print(x)
print(x.shape)
```

```python
#reshaped
x_reshaped = x.reshape(2, 3)

print(x_reshaped)
print(x_reshaped.shape)
```

```python
#PyTorch-এ -1 ব্যবহার করলে PyTorch নিজেই সেই Dimension-এর Size হিসাব করে নেয়।
x = torch.arange(1, 13)

print(x.reshape(3, -1))
```

```python
#view() use  Tensor of shape change
import torch

x = torch.arange(1, 8)

print(x)
print(x.shape)

z = x.view(1, 7)

print(z)
print(z.shape)
```

# stacking - multiple tensor connect together

```python
tensor1 = torch.tensor([1, 2, 3])
tensor2 = torch.tensor([4, 5, 6])

print(tensor1.shape)
print(tensor2.shape)
```

```python
stacked = torch.stack((tensor1, tensor2))

print(stacked)
print(stacked.shape)
```

```python
#torch.cat() আগে থেকে থাকা Dimension বরাবর Tensor যুক্ত করে।
tensor1 = torch.tensor([1, 2, 3])
tensor2 = torch.tensor([4, 5, 6])

# Stack
print(torch.stack((tensor1, tensor2)))

# Concatenate
print(torch.cat((tensor1, tensor2)))
```

# permute() Tensor of dimension or axis of order changes

```python

# Create tensor with specific shape
x_original = torch.rand(size=(224, 224, 3))

# Permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

# Reproducibility (trying to take the random out of random)

```python
import torch

# Create two random tensors
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(f"Tensor A:\n{random_tensor_A}\n")
print(f"Tensor B:\n{random_tensor_B}\n")
print(f"Does Tensor A equal Tensor B? (anywhere)")
random_tensor_A == random_tensor_B
```

 torch.manual_seed(seed). Seed হলো random number generation শুরু করার একটি নির্দিষ্ট starting point।

```python
import torch
import random

# # Set the random seed
RANDOM_SEED=42 # try changing this to different values and see what happens to the numbers below
torch.manual_seed(seed=RANDOM_SEED)
random_tensor_C = torch.rand(3, 4)

# Have to reset the seed every time a new rand() is called
# Without this, tensor_D would be different to tensor_C
torch.random.manual_seed(seed=RANDOM_SEED) # try commenting this line out and seeing what happens
random_tensor_D = torch.rand(3, 4)

print(f"Tensor C:\n{random_tensor_C}\n")
print(f"Tensor D:\n{random_tensor_D}\n")
print(f"Does Tensor C equal Tensor D? (anywhere)")
random_tensor_C == random_tensor_D
```
