```python
#NumPy Array Copy vs View
```

```python
#COPY:
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()
arr[0] = 42

print(arr)
print(x)
```

```python
#VIEW
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
x = arr.view()
arr[0] = 42

print(arr)
print(x)
```

```python
#NumPy Filter Array
```

```python
import numpy as np

# Student names
students = np.array(["Rahim", "Karim", "Sakib", "Nila", "Mina"])

# Student marks
marks = np.array([85, 45, 75, 35, 90])

# 1. Create filtering condition
filter_condition = marks > 50

# 2. Filter student names
filtered_students = students[filter_condition]

# 3. Filter marks
filtered_marks = marks[filter_condition]

# 4. Display results
print("All Students:", students)
print("All Marks:", marks)

print("\nFiltered Students:", filtered_students)
print("Filtered Marks:", filtered_marks)

# 5. Find the index of students with marks above 50
print("\nFiltered Index:", np.where(marks > 50))

# 6. Multiple conditions: marks between 50 and 90
result = marks[(marks > 50) & (marks < 90)]

print("\nMarks between 50 and 90:", result)
```

```python
#Random Numbers in NumPy
```

```python
import numpy as np

# Original Array
arr = np.array([10, 20, 30, 40, 50])

# 1. np.random.rand()
print("1. Random Decimal:")
print(np.random.rand(3))


# 2. np.random.randint()
print("\n2. Random Integer:")
print(np.random.randint(1, 100, size=5))


# 3. np.random.choice()
print("\n3. Random Value Selection:")
print(np.random.choice(arr, size=3))


# 4. np.random.shuffle()
shuffle_arr = arr.copy()
np.random.shuffle(shuffle_arr)

print("\n4. Shuffle Array:")
print("Original:", arr)
print("Shuffled:", shuffle_arr)


# 5. np.random.permutation()
permuted_arr = np.random.permutation(arr)

print("\n5. Permutation:")
print("Original:", arr)
print("New Array:", permuted_arr)


# 6. np.random.seed()
np.random.seed(42)

print("\n6. Random Seed:")
print(np.random.randint(1, 100, size=5))


# 7. np.random.default_rng()
rng = np.random.default_rng(42)

print("\n7. Default RNG:")
print(rng.integers(1, 100, size=5))
```

```python
#Random Data Distribution
```

```python
import numpy as np

# Random Generator
rng = np.random.default_rng(42)

# 1. Normal Distribution
normal = rng.normal(50, 10, 5)

# 2. Binomial Distribution
binomial = rng.binomial(10, 0.5, 5)

# 3. Poisson Distribution
poisson = rng.poisson(5, 5)

# 4. Uniform Distribution
uniform = rng.uniform(10, 50, 5)

# 5. Exponential Distribution
exponential = rng.exponential(5, 5)

# Display Results
print("Normal:", normal)
print("Binomial:", binomial)
print("Poisson:", poisson)
print("Uniform:", uniform)
print("Exponential:", exponential)


import numpy as np

# Random Generator
rng = np.random.default_rng(42)

# 1. Logistic Distribution
logistic = rng.logistic(loc=0, scale=1, size=5)

# 2. Multinomial Distribution
multinomial = rng.multinomial(
    n=100,
    pvals=[0.2, 0.3, 0.5]
)

# 3. Exponential Distribution
exponential = rng.exponential(scale=5, size=5)

# 4. Chi-Square Distribution
chi_square = rng.chisquare(df=3, size=5)

# 5. Rayleigh Distribution
rayleigh = rng.rayleigh(scale=2, size=5)

# 6. Pareto Distribution
pareto = rng.pareto(a=3, size=5)

# 7. Zipf Distribution
zipf = rng.zipf(a=2, size=5)


# Display Results
print("Logistic:", logistic)
print("Multinomial:", multinomial)
print("Exponential:", exponential)
print("Chi-Square:", chi_square)
print("Rayleigh:", rayleigh)
print("Pareto:", pareto)
print("Zipf:", zipf)
```
