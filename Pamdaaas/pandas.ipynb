```python
#map() — প্রতিটি value পরিবর্তন করা
import pandas as pd

df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sadia"],
    "Result": ["Pass", "Fail", "Pass"]
})

df["Result"] = df["Result"].map({
    "Pass": "Successful",
    "Fail": "Unsuccessful"
})

print(df)
```

```python
#replace() — নির্দিষ্ট value বদলানো
df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sadia"],
    "City": ["Dhaka", "DHK", "Chittagong"]
})

df["City"] = df["City"].replace({
    "DHK": "Dhaka",
    "Chittagong": "Chattogram"
})

print(df)
```

```python
#apply() — প্রতিটি value-এর ওপর কাজ চালানো
df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sadia"],
    "Marks": [60, 70, 80]
})

df["New_Marks"] = df["Marks"].apply(lambda x: x + 5)

print(df)
```

```python
#apply() দিয়ে Text পরিবর্তন
df = pd.DataFrame({
    "News": [
        "  Education news  ",
        " Fake job circular ",
        "Sports news   "
    ]
})

df["News"] = df["News"].apply(lambda x: x.strip())

print(df)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["  Rahim  ", "KARIM", "Sadia"],
    "News": [
        "FAKE news about jobs",
        "Real news about education",
        "Fake news about sports"
    ]
})

print(df)
```

```python
#.str.strip() — অতিরিক্ত whitespace সরানো
df["Name"] = df["Name"].str.strip()

print(df["Name"])
```

```python
#Advanced Data Cleaning
#Inconsistent data ঠিক করা

df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sadia"],
    "City": ["Dhaka", "DHAKA", "dhaka"]
})

df["City"] = df["City"].str.strip().str.title()

print(df)
```

```python
#Date & Time — তারিখ নিয়ে কাজ

df = pd.DataFrame({
    "Headline": ["News A", "News B", "News C"],
    "Date": ["2026-01-15", "2026-03-20", "2026-10-05"]
})

print(df.dtypes)
```

```python
#ভুল তারিখ সামলানো
df = pd.DataFrame({
    "Date": ["2026-01-15", "invalid-date", "2026-03-20"]
})

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print(df)
```

```python
#Text পরিষ্কার করেছি।

#Label একই format-এ এনেছি।
#Date-কে datetime-এ রূপান্তর করেছি।


import pandas as pd

df = pd.DataFrame({
    "Headline": [
        "  Fake Job News  ",
        "REAL Education News",
        "Fake Sports News"
    ],
    "Label": ["FAKE", "Real", "fake"],
    "Date": ["2026-01-15", "2026-03-20", "2026-10-05"]
})

# 1. Headline-এর whitespace সরানো
df["Headline"] = df["Headline"].str.strip()

# 2. Label standardize করা
df["Label"] = df["Label"].str.strip().str.lower()

# 3. Date-কে datetime-এ রূপান্তর
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# 4. শুধু fake label-এর row বের করা
fake_news = df[df["Label"] == "fake"]

print(fake_news)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Rahim", "Karim", "Sadia", "Nila", "Hasan",
             "Rima", "Jamal", "Mina", "Arif", "Lima"],

    "Department": ["CSE", "EEE", "CSE", "BBA", "EEE",
                   "CSE", "BBA", "EEE", "CSE", "BBA"],

    "Study_Hours": [5, 3, 7, 2, 4, 6, 3, 5, 8, 4],

    "Attendance": [90, 75, 95, 60, 80, 85, 70, 88, 98, 78],

    "Marks": [85, 65, 95, 45, 70, 88, 55, 80, 98, 72]
})

print(df)
```

```python
#Step 1 — Understand the Dataset
df.head()
df.tail()
print(df.shape)
df.info()
df.describe()
print(df.columns)
print(df.dtypes)
```

```python
#Step 2 — Missing Value Analysis
df.loc[2, "Attendance"] = np.nan
df.loc[5, "Marks"] = np.nan

print(df.isnull().sum())
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
#tep 3 — Duplicate Data Analysis
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)
print(df.duplicated().sum())
print(df[df.duplicated()])
df = df.drop_duplicates()

print(df.shape)
```

```python
#Step 4 — Univariate Analysis
print("Mean:", df["Marks"].mean())
print("Median:", df["Marks"].median())
print("Minimum:", df["Marks"].min())
print("Maximum:", df["Marks"].max())

print(df["Department"].value_counts())

df["Marks"].plot(kind="hist", bins=5)

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()


df["Marks"].plot(kind="box")

plt.title("Marks Boxplot")
plt.ylabel("Marks")

plt.show()
```

```python
#Step 5 — Outlier Detection

marks = pd.Series([60, 65, 70, 75, 80, 85, 90, 300])

Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = marks[
    (marks < lower) | (marks > upper)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:")
print(outliers)
```

```python
#Step 6 — Bivariate Analysis
df.plot(
    x="Study_Hours",
    y="Marks",
    kind="scatter"
)

plt.title("Study Hours vs Marks")
plt.show()

print(df[["Study_Hours", "Attendance", "Marks"]].corr())

result = df.groupby("Department")["Marks"].mean()

print(result)
```

```python
#Step 7 — Correlation Heatmap

import seaborn as sns
import matplotlib.pyplot as plt

corr = df[[
    "Study_Hours",
    "Attendance",
    "Marks"
]].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
import pandas as pd

data = {
    "Name": ["Rahim", "Karim", "Sakib", "Nila", "Mitu", "Riya"],
    "Department": ["CSE", "CSE", "EEE", "EEE", "CSE", "EEE"],
    "Marks": [85, 70, 90, 60, 95, 75],
    "Age": [21, 22, 21, 23, 22, 21]
}

df = pd.DataFrame(data)

print(df)
```

```python
#groupby() — Group অনুযায়ী হিসাব
df.groupby("Department")["Marks"].mean()
```

```python
#transform() — Group-এর হিসাব প্রতিটি Row-তে দেখানো
df.groupby("Department")["Marks"].mean()
```

```python
#rolling() — Moving Average
sales = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [100, 120, 150, 130, 170]
})

sales["Moving_Avg"] = (
    sales["Sales"].rolling(window=3).mean()
)

print(sales)
```

```python
#rank() — Ranking তৈরি করা
df["Rank"] = df["Marks"].rank(
    ascending=False,
    method="dense"
)

print(df[["Name", "Marks", "Rank"]])
```

```python
#qcut() — সমান সংখ্যক Data-তে ভাগ করা
df["Performance_Group"] = pd.qcut(
    df["Marks"],
    q=3,
    labels=["Low", "Medium", "High"]
)

print(df[["Name", "Marks", "Performance_Group"]])
```
